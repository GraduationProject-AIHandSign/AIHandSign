```python
# 2. GCS 연결 확인
from google.cloud import storage

PROJECT_ID = "project-id"  # 본인 프로젝트 ID
BUCKET_NAME = "sign-language-data-2026"

client = storage.Client(project=PROJECT_ID)
bucket = client.bucket(BUCKET_NAME)

# 버킷 폴더 목록 출력
blobs = bucket.list_blobs()
for blob in blobs:
    print(blob.name)

print("GCS 연결 성공")
```
